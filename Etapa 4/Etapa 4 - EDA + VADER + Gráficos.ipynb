## Iniciando script de análise de sentimento

```python
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer

nltk.download('vader_lexicon')
```

```python
# Inicializa o analisador de sentimentos
sia = SentimentIntensityAnalyzer()

def classificar_sentimento(texto):
    """
    Função que recebe um texto, calcula o 'compound score' usando o VADER
    e retorna a classificação (POSITIVO, NEUTRO, NEGATIVO).
    """
    # Trata valores nulos/vazios caso existam no CSV
    if pd.isna(texto):
        return 'NEUTRO'
    
    # Calcula a polaridade. Retorna um dicionário e extraímos a chave 'compound'
    score = sia.polarity_scores(str(texto))['compound']
    
    # Aplica as regras de negócio de acordo com a pontuação
    if score >= 0.05:
        return 'POSITIVO'
    elif score <= -0.05:
        return 'NEGATIVO'
    else:
        return 'NEUTRO'
```

```python
# Calculando o sentimento de cada parte
data['headline_sentimento'] = data['review_headline'].apply(classificar_sentimento)
data['body_sentimento'] = data['review_body'].apply(classificar_sentimento)

# Criando uma coluna para analise do sentimento geral, combinando headline e body
data['review_completa'] = data['review_headline'].fillna('') + ". " + data['review_body'].fillna('')
data['sentimento_geral'] = data['review_completa'].apply(classificar_sentimento)
```

```python
# Configurar o estilo visual do Seaborn
sns.set_theme(style="whitegrid")
# Definir uma paleta de cores padrão para manter consistência
cores = {'POSITIVO': '#2ecc71', 'NEUTRO': '#95a5a6', 'NEGATIVO': '#e74c3c'}
ordem_categorias = ['POSITIVO', 'NEUTRO', 'NEGATIVO']

# Criar uma figura com 3 subgráficos (1 linha, 3 colunas)
fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)
fig.suptitle('Análise de Sentimentos das Avaliações (VADER)', fontsize=20, fontweight='bold')

# Gráfico 1: Headline (Título)
sns.countplot(
    data=data, 
    x='headline_sentimento', 
    ax=axes[0], 
    order=ordem_categorias, 
    palette=cores
)
axes[0].set_title('Sentimento do Título (Headline)', fontsize=14)
axes[0].set_xlabel('Classificação')
axes[0].set_ylabel('Quantidade de Avaliações')

# Gráfico 2: Body (Corpo)
sns.countplot(
    data=data, 
    x='body_sentimento', 
    ax=axes[1], 
    order=ordem_categorias, 
    palette=cores
)
axes[1].set_title('Sentimento do Texto (Body)', fontsize=14)
axes[1].set_xlabel('Classificação')
axes[1].set_ylabel('')

# Gráfico 3: Sentimento Geral
sns.countplot(
    data=data, 
    x='sentimento_geral', 
    ax=axes[2], 
    order=ordem_categorias, 
    palette=cores
)
axes[2].set_title('Sentimento Geral', fontsize=14)
axes[2].set_xlabel('Classificação')
axes[2].set_ylabel('')

# Adicionar os rótulos de dados (valores em cima de cada barra)
for ax in axes:
    for p in ax.patches:
        height = p.get_height()
        if height > 0: # Para não mostrar 0 em barras vazias
            ax.annotate(f'{int(height)}', 
                        (p.get_x() + p.get_width() / 2., height), 
                        ha='center', va='bottom', 
                        fontsize=12, fontweight='bold', color='black', 
                        xytext=(0, 5), textcoords='offset points')

# Ajustar o layout para não cortar textos
plt.tight_layout(rect=[0, 0, 1, 0.95])

# Salvar o gráfico como imagem e exibi-lo
nome_imagem = 'resultado_sentimentos.png'
plt.savefig(nome_imagem, dpi=300)

plt.show()
```

```python
# Calculo de acurácia do modelo utilizando o star_rating como base de comparação
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# Criando o gabarito baseado nas estrelas
def mapear_estrelas_para_sentimento(nota):
    if nota >= 4: return 'POSITIVO'
    elif nota <= 3: return 'NEGATIVO'
    else: return 'NEUTRO'

data['sentimento_real'] = data['star_rating'].apply(mapear_estrelas_para_sentimento)

# Calculando a Acurácia do 'body_sentimento' contra o gabarito
acuracia = accuracy_score(data['sentimento_real'], data['body_sentimento'])
print(f"Acurácia do VADER (Body): {acuracia:.2%}")

# Gerando o relatório detalhado
print("\nRelatório de Classificação:")
print(classification_report(data['sentimento_real'], data['body_sentimento']))
```

```python
# Calculando a Acurácia do 'headline_sentimento' contra o gabarito
acuracia = accuracy_score(data['sentimento_real'], data['headline_sentimento'])
print(f"Acurácia do VADER (Body): {acuracia:.2%}")

# Gerando o relatório detalhado
print("\nRelatório de Classificação:")
print(classification_report(data['sentimento_real'], data['headline_sentimento']))
```

```python
# Calculando a Acurácia do 'sentimento_geral' contra o gabarito
acuracia = accuracy_score(data['sentimento_real'], data['sentimento_geral'])
print(f"Acurácia do VADER (Body): {acuracia:.2%}")

# Gerando o relatório detalhado
print("\nRelatório de Classificação:")
print(classification_report(data['sentimento_real'], data['sentimento_geral']))
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# 1. Garantir que a data está no formato correto e criar a coluna ano_mes
data['review_date'] = pd.to_datetime(data['review_date'], errors='coerce')
data['ano_mes'] = data['review_date'].dt.to_period('M')

# 2. Agrupar os dados calculando a MÉDIA das notas e a CONTAGEM de avaliações (volume)
analise_temporal = data.groupby('ano_mes').agg(
    nota_media=('star_rating', 'mean'),
    volume=('star_rating', 'count') 
).reset_index()

# 3. Converter 'ano_mes' para timestamp para o Matplotlib aceitar no eixo X
analise_temporal['ano_mes'] = analise_temporal['ano_mes'].dt.to_timestamp()

# 4. Configurar a figura e o Eixo Principal (Esquerdo) para a Nota Média
fig, ax1 = plt.subplots(figsize=(14, 6))

cor_nota = '#1f77b4' # Azul
ax1.plot(analise_temporal['ano_mes'], analise_temporal['nota_media'], marker='o', linestyle='-', color=cor_nota, linewidth=2)
ax1.set_xlabel('Data da Avaliação', fontsize=12)
ax1.set_ylabel('Nota Média (Estrelas)', fontsize=12, color=cor_nota, fontweight='bold')
ax1.tick_params(axis='y', labelcolor=cor_nota)

# Formatando as datas no eixo X
ax1.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
ax1.xaxis.set_major_locator(mdates.AutoDateLocator())
ax1.tick_params(axis='x', rotation=45)
ax1.grid(True, linestyle='--', alpha=0.5) # Grid suave

# 5. Configurar o Eixo Secundário (Direito) para o Volume de Vendas/Avaliações
ax2 = ax1.twinx() # Cria um segundo eixo Y que compartilha o mesmo eixo X

cor_volume = '#ff7f0e' # Laranja
# Usamos um gráfico de barras com transparência (alpha=0.3) para ficar no fundo
ax2.bar(analise_temporal['ano_mes'], analise_temporal['volume'], width=20, alpha=0.7, color=cor_volume)
ax2.set_ylabel('Volume de Avaliações (Qtd)', fontsize=12, color=cor_volume, fontweight='bold')
ax2.tick_params(axis='y', labelcolor=cor_volume)

# 6. Adicionar título geral
plt.title('Evolução Temporal: Nota Média vs. Volume de Avaliações', fontsize=15, pad=15)

# Ajustar o layout para nada ficar cortado e exibir
fig.tight_layout()
plt.show()
```

```python
# 1. Calculando as métricas
# Contagem total de avaliações por status
contagem = data['verified_purchase'].value_counts()

# Média de notas
medias = data.groupby('verified_purchase')['star_rating'].mean()

# Distribuição percentual
distribuicao = data.groupby(['verified_purchase', 'star_rating']).size().unstack()
distribuicao_pct = distribuicao.div(distribuicao.sum(axis=1), axis=0) * 100

# 2. Configurando a área de plotagem (agora 1 linha, 3 colunas)
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(20, 6))

# --- GRÁFICO 1: Volume Absoluto (NOVO) ---
# Usando cores consistentes: azul para 'Y' e vermelho/salmão para 'N'
cores_volume = ['#66b3ff' if i == 'Y' else '#ff9999' for i in contagem.index]
contagem.plot(kind='bar', ax=ax1, color=cores_volume, edgecolor='black')
ax1.set_title('Volume Total de Avaliações', fontsize=14, pad=15)
ax1.set_xlabel('Compra Verificada', fontsize=12)
ax1.set_ylabel('Quantidade Absoluta', fontsize=12)
ax1.set_xticklabels(contagem.index, rotation=0)
ax1.grid(axis='y', linestyle='--', alpha=0.7)

# Adicionando os valores exatos no topo das barras do volume
for i, valor in enumerate(contagem):
    ax1.text(i, valor + (valor * 0.02), f"{valor:,}", ha='center', fontsize=12, fontweight='bold')

# --- GRÁFICO 2: Distribuição das Notas (Proporção %) ---
distribuicao_pct.T.plot(kind='bar', ax=ax2, color=['#ff9999', '#66b3ff'], edgecolor='black', width=0.7)
ax2.set_title('Distribuição das Notas (Proporção %)', fontsize=14, pad=15)
ax2.set_xlabel('Nota (Star Rating)', fontsize=12)
ax2.set_ylabel('Porcentagem dentro do grupo (%)', fontsize=12)
ax2.legend(title='Compra Verificada?', labels=['Não (N)', 'Sim (Y)'])
ax2.grid(axis='y', linestyle='--', alpha=0.7)
ax2.tick_params(axis='x', rotation=0)

# --- GRÁFICO 3: Nota Média ---
# Garantindo que a ordem seja ['N', 'Y'] para as cores baterem com a legenda
medias = medias.reindex(['N', 'Y']) 
medias.plot(kind='bar', ax=ax3, color=['#ff9999', '#66b3ff'], edgecolor='black')
ax3.set_title('Nota Média: Verificada vs Não Verificada', fontsize=14, pad=15)
ax3.set_xlabel('Compra Verificada', fontsize=12)
ax3.set_ylabel('Nota Média (Estrelas)', fontsize=12)
ax3.set_xticklabels(['Não (N)', 'Sim (Y)'], rotation=0)
ax3.grid(axis='y', linestyle='--', alpha=0.7)

# Adicionando os valores numéricos exatamente em cima das barras de média
for i, valor in enumerate(medias):
    if pd.notna(valor): # Checagem de segurança
        ax3.text(i, valor + 0.05, f"{valor:.2f}", ha='center', fontsize=12, fontweight='bold')

# Ajusta os espaçamentos e exibe
plt.tight_layout()
plt.show()
```

```python
# 1. Criar uma função para agrupar as notas reais nas duas faixas de interesse
def classificar_faixa(nota):
    if nota <= 3.0:
        return 'Notas Baixas (1 a 3)'
    elif nota >= 4.0:
        return 'Notas Altas (4 a 5.0)'
    else:
        return 'Notas Médias (3 a 3.99)'

# Criando a nova coluna de classificação
data['faixa_analise'] = data['star_rating'].apply(classificar_faixa)

# 2. Filtrar o DataFrame para ignorar as notas médias, focando apenas nos extremos
df_foco = data[data['faixa_analise'].isin(['Notas Baixas (1 a 3)', 'Notas Altas (4 a 5.0)'])]

# 3. Criar uma tabela cruzada (cruzando a Faixa da Nota com o Sentimento do VADER)
tabela_erros = pd.crosstab(df_foco['faixa_analise'], df_foco['sentimento_geral'])

# 4. Configurar e plotar o gráfico de barras agrupadas
ax = tabela_erros.plot(kind='bar', figsize=(12, 6), edgecolor='black', width=0.7)

# Títulos e rótulos
plt.title('Análise de Erros do VADER (Falsos Positivos vs Falsos Negativos)', fontsize=15, pad=20)
plt.xlabel('Realidade do Cliente (Nota Real)', fontsize=12)
plt.ylabel('Número de Avaliações', fontsize=12)

# Adicionando os valores numéricos exatamente em cima das barras para clareza
for p in ax.patches:
    valor = int(p.get_height())
    if valor > 0: # Mostra o número apenas se a barra existir (maior que zero)
        ax.annotate(f"{valor}", 
                    (p.get_x() + p.get_width() / 2., p.get_height()), 
                    ha='center', va='bottom', xytext=(0, 5), 
                    textcoords='offset points', fontsize=11, fontweight='bold')

# Configurando o visual do eixo X para as legendas ficarem retas
plt.xticks(rotation=0, fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)

# Configurando a legenda
plt.legend(title='VADER classificou como:', title_fontsize='12', fontsize='11')

# Imprimindo um texto de ajuda no console para facilitar a leitura
print("-" * 60)
print("📌 GUIA DE LEITURA DO GRÁFICO:")
print("1. Em 'Notas Baixas', observe a barra de sentimento POSITIVO.")
print("   -> Estes são os FALSOS POSITIVOS (VADER achou bom, mas a nota era terrível).")
print("2. Em 'Notas Altas', observe a barra de sentimento NEGATIVO.")
print("   -> Estes são os FALSOS NEGATIVOS (VADER achou ruim, mas o cliente amou).")
print("-" * 60)

# Ajuste fino e exibição
plt.tight_layout()
plt.show()
```
